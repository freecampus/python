## 1. Give each clue position one fixed meaning

A treasure scanner produces a row, a column, and a symbol together:

````python
clue = (12, 7, "N")

print(clue)
print(type(clue).__name__)
print(len(clue))
````

This tuple is an ordered three-position record. Position 0 always means row,
position 1 column, and position 2 symbol. Unlike an adventure backpack, the clue
is not intended to gain or lose positions.

Tuples work well for small values whose **length and positional meaning are
fixed**: coordinates, RGB colors, database result rows, or the pair returned by
`divmod()`. A dictionary is usually clearer when a record has many optional or
self-describing fields.

This lesson answers four practical questions:

- What syntax creates zero-, one-, and many-item tuples?
- Which sequence operations still work when positions cannot be changed?
- How does unpacking make positional meaning visible through names?
- When should a variable-length remainder use one starred target?

## 2. The comma creates the tuple

Parentheses make tuple boundaries easy to see, but the comma is the decisive
syntax:

````python
parenthesized = (12, 7)
packed = 12, 7

assert parenthesized == packed
assert type(packed) is tuple
````

This difference matters for a one-item tuple:

````python
not_a_tuple = (5)
one_item_tuple = (5,)

print(type(not_a_tuple).__name__)
print(type(one_item_tuple).__name__)
````

`(5)` merely groups the integer expression. `(5,)` is a tuple containing one
integer. A trailing comma is also accepted in longer tuples and makes multi-line
editing safer:

````python
checkpoint = (
    "garden",
    12,
    True,
)
````

The empty tuple needs no comma:

````python
empty = ()
also_empty = tuple()

assert empty == also_empty
````

The `tuple()` constructor consumes values from another iterable:

````python
letters = tuple("NOVA")
numbers = tuple([3, 1, 4])

print(letters)
print(numbers)
````

The string supplies individual characters; the list supplies its integer items.

## 3. Read a tuple like any ordered sequence

Tuples support length, indexing, negative indexing, slicing, membership, counting,
and first-match search:

````python
clue = (12, 7, "N", "N")

print(len(clue))
print(clue[0])
print(clue[-1])
print(clue[1:3])
print("N" in clue)
print(clue.count("N"))
print(clue.index("N"))
````

`clue[1:3]` is a new tuple because slicing preserves the sequence family. The
count is two, and `index("N")` returns only the first matching position, index 2.

As with a list, an absent numeric position raises `IndexError` and a missing value
passed to `.index()` raises `ValueError`.

### Tuple positions cannot be replaced

````python
coordinate = (12, 7)
coordinate[0] = 13
````

Python raises `TypeError: 'tuple' object does not support item assignment`. Tuples
also have no `.append()`, `.remove()`, or `.clear()` methods. To represent a
different coordinate, create a different tuple:

````python
coordinate = (12, 7)
moved_coordinate = (coordinate[0] + 1, coordinate[1])

print(coordinate)
print(moved_coordinate)
````

The original remains `(12, 7)`.

### An immutable tuple can point to a mutable value

````python
mission = ("NOVA", ["map", "rope"])
mission[1].append("torch")

print(mission)
````

The tuple still has the same two positions and still refers to the same list in
position 1. The nested list itself permits mutation. “A tuple is immutable” means
its own sequence of references cannot be replaced, added, or removed; it does not
promise that every reachable object is frozen. Unit 6 explores that object graph
in depth.

## 4. Unpack fixed positions into meaningful names

Unpacking assigns each supplied position to a target name:

````python
clue = (12, 7, "N")
row, column, symbol = clue

print(row)
print(column)
print(symbol)
````

Read the assignment in two stages. Python evaluates the right side first and
obtains three values. It then assigns the first to `row`, the second to `column`,
and the third to `symbol`.

The names expose meaning that `clue[0]`, `clue[1]`, and `clue[2]` hide:

````python
is_expected_clue = row == 12 and column == 7 and symbol == "N"
print(is_expected_clue)
````

Parentheses around targets are optional:

````python
(red, green, blue) = (40, 80, 120)
````

Use them when they clarify a nested record boundary, not because unpacking
requires them.

### Target count must match value count

````python
row, column = (12, 7, "N")
````

Python raises `ValueError: too many values to unpack (expected 2)`.

````python
row, column, symbol = (12, 7)
````

This raises `ValueError: not enough values to unpack`. These errors are evidence
that the supplied record shape disagrees with the assignment’s promised shape.
Do not silence them by adding meaningless throwaway names until you understand
which data contract changed.

### Swap names without a temporary variable

````python
left = "map"
right = "key"

left, right = right, left

print(left)
print(right)
````

Python first packs the right-side values, then unpacks them into the left-side
names. `left` becomes `"key"` and `right` becomes `"map"` without losing either
original value.

### Unpack values returned together

`divmod()` returns quotient and remainder as a two-item tuple:

````python
complete_crates, leftover_items = divmod(29, 6)

assert complete_crates == 4
assert leftover_items == 5
````

Naming the two meanings immediately is clearer than retaining a value such as
`result[0]` and `result[1]`.

## 5. One starred target absorbs a variable-length remainder

When the first and last positions have fixed meanings but the middle varies, use
one starred target:

````python
first, *middle, last = ("N", "O", "V", "A")

print(first)
print(middle)
print(last)
print(type(middle).__name__)
````

`first` is `"N"`, `last` is `"A"`, and `middle` is the **list** `["O", "V"]`.
The starred target always collects its share into a list, even when the source is
a tuple.

It can appear at the beginning or end:

````python
*earlier, latest = (4, 7, 9, 12)
leader, *followers = ("Nova", "Mira", "Sol")

print(earlier)
print(latest)
print(leader)
print(followers)
````

One assignment may contain only one starred target because two stars would make
the division ambiguous:

```text
*beginning, *ending = (1, 2, 3, 4)
```

The syntax is invalid before the program runs.

An empty remainder is valid:

````python
first, *middle, last = ("N", "A")
assert middle == []
````

There are still enough values for the two unstarred targets.

## 6. Choose tuple or list from the promised lifecycle

Both types preserve order, allow duplicates, and support indexes and slices. The
difference is what the program promises to do next:

| Requirement | Prefer | Reason |
|---|---|---|
| Ordered collection that grows or shrinks | `list` | Its sequence structure is mutable. |
| Small fixed positional record | `tuple` | Its length and positions cannot be changed. |
| Named record fields | `dict` | Keys communicate meaning better than distant numeric positions. |
| Unique membership with no positions | `set` | Uniqueness and group comparisons are its purpose. |

Do not choose a tuple merely to prevent every possible change. A tuple containing
mutable objects is not deeply frozen, and a value often needs validation beyond
its container type. Choose it when fixed positional shape communicates the model.

### Hashability is a practical preview

Some tuples can be dictionary keys or set members:

````python
visited_coordinates = {(12, 7), (5, 3)}
terrain_by_coordinate = {(12, 7): "vault", (5, 3): "garden"}

print((12, 7) in visited_coordinates)
print(terrain_by_coordinate[(5, 3)])
````

A tuple is hashable only when all values needed for equality are themselves
hashable. A tuple containing a list is not:

````python
invalid_key = ("NOVA", [12, 7])
catalog = {invalid_key: "signal"}
````

This raises `TypeError: unhashable type: 'list'`. Lessons 3 and 4 use the rule;
Unit 6 explains hashability systematically.

## Key points

> **Key points**
- The comma creates a tuple; parentheses normally clarify its boundary.
- Tuples preserve order and duplicates but do not permit replacing, inserting, or
  deleting their own positions.
- A tuple may still contain a mutable object.
- Ordinary unpacking requires matching value and target counts; one starred target
  absorbs a variable-length remainder into a list.
- Fixed, compact positional records often suit tuples; changing sequences suit
  lists, and named records often suit dictionaries.
- A tuple can be a dictionary key or set member only when its relevant contents
  are hashable.
