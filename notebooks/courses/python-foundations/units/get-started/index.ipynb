# Get Started Overview

This notebook was generated from the FreeCampus Python lesson source. Run cells from top to bottom, write predictions before execution, and change one thing at a time.

Source lesson: `courses/python-foundations/units/get-started/index.qmd`

- **Level:** Python Foundations · Unit 0
- **Estimated time:** 13–20 hours including the challenge
- **Unit outcome:** Run and explain a small Python program, keep trustworthy evidence in a clean notebook, choose a suitable practice environment, diagnose common first-run problems, and use AI assistance without surrendering prediction or verification.
- **Practice in:** Google Colab or Jupyter, then a local editor and terminal when your device permits

## 1. Start with a working place to learn

You do not need programming experience before this unit. You do not need to
memorize Python words, choose a career specialty, or install every tool you have
heard about. You need one place where you can run a small program, observe what
happens, preserve useful notes, and try again without hiding the evidence.

The fastest route is a browser notebook such as Google Colab. It lets you begin
without installing Python. A local workspace becomes important when you need to
work with saved files, project folders, terminals, editors, and isolated
environments. This unit teaches both routes. If a school or work computer blocks
installation, you will record that constraint, continue in Colab, and identify
what access you will need before later project units.

Notice that both routes lead to the same learning loop and a clean rerun.

A browser-first learner may keep the same material in a Colab notebook and
Google Drive, then download a copy. A local learner creates the folder on their
computer. The filenames are less important than what they prove:

- `first_run.py` contains a complete program whose three output lines you can
  predict and explain;
- `learning-log.md` preserves a prediction, observed result, explanation,
  controlled change, and repaired failure;
- `first_notebook.ipynb` combines readable notes with code and succeeds after a
  clean restart and top-to-bottom run; and
- `launch-card.md` records the environment, working command, project location,
  interpreter evidence, fallback, and help route you can depend on.

These are not polished portfolio projects. They are the workshop bench on which
you will build later projects.

## 4. Prepare without buying anything

Bring:

- a modern web browser;
- a place to save a notebook or download files;
- permission to create a small folder if you will practise locally; and
- a plain-text note for commands, outputs, and questions.

The course does not require a paid editor, notebook service, or AI subscription.
You can complete the AI lesson using the supplied responses. Do not paste
passwords, API keys, private repository code, personal records, or assessment
answers into an external service.

> **Choose one route before adding tools**
Begin in Colab if installation would delay your first run. Begin locally if
Python is already available and you can create files. Do not install Python,
Conda, Anaconda, multiple editors, and several notebook systems at once. One
working route creates better evidence than five half-configured routes.

## 5. Plan realistic stopping points

The estimate is active work rather than silent reading:

- **First program:** about 2–3 hours, including predictions, repair cases,
  checkpoints, and the final script.
- **Learning with evidence:** about 2.5–3.5 hours, including a complete debugging
  record.
- **Responsible AI assistance:** about 2–3 hours, with no live AI account
  required.
- **Notebook workflow:** about 2.5–4 hours, including a clean restart and run.
- **Local workspace:** about 3–5 hours. Installation and platform recovery vary
  by device.
- **Launch challenge:** about 35–50 minutes.

A good stopping point is the end of a final lab. Save the artifact and its exact
observed result. When you return, rerun the nearest check before assuming the
old environment still matches your notes.

## 6. Know what success looks like

You are ready for Unit 1 when you can:

1. point to the source line responsible for each line of a tiny program's
   output;
2. keep prediction, observation, explanation, and next change separate;
3. treat an AI response as a claim to verify rather than permission to stop
   thinking;
4. restart a notebook and run every cell from the top without relying on hidden
   state;
5. state which Python command and environment you are using; and
6. preserve the exact command, output, and relevant file when you need help.

Progress in this site is local and self-reported. It is not a verified exam,
identity check, or certificate record. A Foundations final project and
certificate are planned but are not available yet; they will be announced
later.
