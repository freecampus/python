## 1. One statement can occupy several screen lines

A **physical line** is one line visible in the editor. A **logical line** is the
complete instruction Python reads as one statement.

This assignment fits on one physical line:

````python
event_total = room_cost + materials_cost + refreshment_cost + equipment_cost
````

The same logical statement can use several physical lines:

````python
event_total = (
    room_cost
    + materials_cost
    + refreshment_cost
    + equipment_cost
)
````

The opening `(` tells Python the expression is not finished. Newlines inside the
matching parentheses do not end the assignment.

Run a complete example:

````python
room_cost = 120
materials_cost = 80
refreshment_cost = 45
equipment_cost = 30

event_total = (
    room_cost
    + materials_cost
    + refreshment_cost
    + equipment_cost
)

print(event_total)
````

The output is `275`, exactly as it would be with a one-line expression.

### What Python does on each line

1. The four input assignments run normally.
2. `event_total = (` starts an assignment whose right side is still open.
3. Python collects the following name and operator tokens.
4. The closing `)` completes the grouped expression.
5. Python evaluates the complete addition.
6. Assignment binds `event_total` to `275`.
7. The final call displays it.

The blank lines improve visual grouping. They do not create or end a block.

Several physical lines can form one logical statement while delimiters remain
open.

## 2. Open delimiters permit safe continuation

Python supports implicit continuation inside three delimiter pairs:

| Pair | Familiar use | Example |
|---|---|---|
| `(` and `)` | grouped expression or function call | `round(total, 2)` |
| `[` and `]` | list or indexing | `[25, 30, 45]` |
| `{` and `}` | dictionary or set | `{"course": "Python"}` |

Collections receive full coverage in Unit 3. Here, focus on the opening and
closing shape.

### Group a calculation

````python
active_minutes = (
    session_count * session_minutes
    - break_count * break_minutes
)
````

The outer parentheses keep the expression open. The two physical calculation
lines form one value.

### Spread call arguments vertically

````python
print(
    "Course:",
    course_name,
    "Minutes:",
    active_minutes,
)
````

Each comma separates an argument. The final comma after `active_minutes` is a
**trailing comma**. Python accepts it, and it makes adding or rearranging lines
easier.

### Lay out a collection as a preview

````python
session_lengths = [
    25,
    45,
    30,
]
````

The list still contains three values. The layout gives each item a clear line
and leaves a trailing comma after the last item.

````python
print(session_lengths)
````

Expected output:

```text
[25, 45, 30]
```

### A closing delimiter must match the opening one

This is invalid:

````python
session_lengths = [
    25,
    45,
    30,
)
````

The list opens with `[` but closes with `)`. Repair the final symbol to `]`.

> **Pair delimiters before reading the details**
When a long statement looks confusing, first match every `(` with `)`, `[` with
`]`, and `{` with `}`. Only then trace the values and operators inside.

## 3. Continuation indentation is not a code block

Compare a block:

````python
if session_minutes >= 30:
    message = "Focused session"
````

with a continued expression:

````python
message = (
    "Focused "
    + "session"
)
````

The indented block line belongs to the `if` header because the header ends in a
colon. The indented string lines belong to an expression whose parenthesis is
still open.

Both use visual indentation, but the grammar is different:

| Question | Block | Continued statement |
|---|---|---|
| What starts it? | A header ending in `:` | An unclosed `(`, `[`, or `{` |
| What ends it? | Dedenting to an earlier level | The matching closing delimiter |
| Does indentation control behavior? | Yes | Usually no; delimiters control continuation |
| Typical style | Four spaces per block level | Align for readability |

### Continuation inside a block

The two forms can appear together:

````python
if session_minutes >= 30:
    message = (
        course_name
        + ": focused session"
    )
````

Read the left edge:

1. `if ...:` begins a block.
2. `message = (` is four spaces inside that block.
3. The continued expression is indented farther for readability.
4. The closing parenthesis lines up with the start of the continued expression.
5. Dedenting to the left margin would end the `if` block.

You do not need to master the condition yet. Your task is to distinguish the
block level from the continuation layout.

## 4. Choose a layout that exposes the structure

Python accepts several layouts inside parentheses. People still need a
consistent visual pattern.

A compact call is clear when it is short:

````python
rounded_total = round(event_total, 2)
````

A vertical call is clearer when arguments need separate attention:

````python
print(
    "Event:",
    event_name,
    "Learners:",
    learner_count,
    "Total:",
    event_total,
)
````

Avoid a half-expanded shape:

````python
print("Event:", event_name,
    "Learners:", learner_count,
    "Total:", event_total)
````

It may run, but the argument boundaries are harder to scan and the indentation
does not reveal a deliberate structure.

Prefer either one clear line or one argument per line:

````python
print(
    "Event:",
    event_name,
    "Learners:",
    learner_count,
    "Total:",
    event_total,
)
````

### Break before operators consistently

This course uses operators at the beginning of continuation lines:

````python
net_total = (
    ticket_income
    + sponsor_income
    - room_cost
    - material_cost
)
````

The vertical operator column makes the additions and subtractions visible.

Do not alternate styles inside one expression:

````python
net_total = (
    ticket_income +
    sponsor_income
    - room_cost -
    material_cost
)
````

The code can be valid, but a consistent pattern is easier to review.

### Use meaningful subexpressions

A very long continued expression can still hide its meaning:

````python
amount_due = (
    ticket_price * adult_count
    + ticket_price * 0.5 * child_count
    + 12
    - 20
)
````

Named stages may be better:

````python
adult_total = ticket_price * adult_count
child_total = ticket_price * 0.5 * child_count
booking_fee = 12
credit = 20

amount_due = (
    adult_total
    + child_total
    + booking_fee
    - credit
)
````

Multiline layout solves width. Intermediate names solve meaning. Use both when
the program benefits from both.

## 5. Trailing commas support safer edits

Compare a vertical call without a trailing comma:

````python
print(
    course_name,
    focused_minutes
)
````

and with one:

````python
print(
    course_name,
    focused_minutes,
)
````

Both run. The trailing comma makes future editing simpler: a new argument can be
added without modifying the previous line.

It also tells formatters that the structure may remain expanded. Formatting
tools are covered later; the beginner habit is simple:

> In a structure with one item per line, include a trailing comma after the last
> item when Python permits it.

### A one-item tuple is a special case

This is a preview of Unit 3:

````python
one_score = (90,)
print(one_score)
````

The comma, not the parentheses alone, creates the one-item tuple. Compare:

````python
grouped_score = (90)
print(grouped_score)
````

`grouped_score` is just the number `90` grouped in parentheses.

Do not memorize all tuple behavior now. Remember that punctuation can carry
meaning beyond visual layout.

## 6. Long text needs a deliberate strategy

This short text fits on one line:

````python
message = "Python Foundations begins with careful reading."
````

Adjacent string literals inside parentheses are joined by Python:

````python
message = (
    "Python Foundations begins with careful reading. "
    "Small experiments turn syntax rules into practical skill."
)

print(message)
````

The space after `reading.` is inside the first string. Without it, the words
would touch.

For visible line breaks inside the text, use an escape sequence:

````python
message = "First session\nSecond session"
print(message)
````

Expected output:

```text
First session
Second session
```

Triple-quoted strings can contain physical newlines:

````python
message = """First session
Second session"""
print(message)
````

Triple-quoted strings also have a special role as docstrings when they appear in
particular positions. The next lesson explains that difference.

### Do not place a bare newline inside ordinary quotes

This is invalid:

````python
message = "First session
Second session"
````

An ordinary quoted string must close before the physical line ends. Use `\n`,
adjacent literals in parentheses, or a triple-quoted string according to the
desired value.

## 7. Avoid fragile continuation with backslashes

Python permits an explicit backslash at the end of a line:

````python
event_total = room_cost + materials_cost + \
    refreshment_cost
````

It works, but it is fragile. A space or comment after the backslash breaks the
continuation, and editing delimiter-based code is usually safer.

Prefer:

````python
event_total = (
    room_cost
    + materials_cost
    + refreshment_cost
)
````

Do not write a comment after a continuation backslash:

````python
event_total = room_cost + materials_cost + \  # more costs below
    refreshment_cost
````

The backslash must be the final character on its physical line. The better
repair is to use parentheses, not merely delete the comment.

> **Default to delimiters**
Use implicit continuation inside `()`, `[]`, or `{}`. Treat backslash
continuation as syntax you may need to recognize, not the normal style to write.

## 8. Put one simple statement on each line

Python permits semicolons between simple statements:

````python
course = "Python"; minutes = 45; print(course, minutes)
````

The line runs, but three actions are hidden on one physical line. Prefer:

````python
course = "Python"
minutes = 45
print(course, minutes)
````

A semicolon does not help with a complex statement that owns an indented body.
This is invalid and unreadable:

````python
if minutes >= 30: message = "focused"; else: message = "short"
````

Use visible blocks:

````python
if minutes >= 30:
    message = "focused"
else:
    message = "short"
````

The decision behavior is still a preview. The syntax lesson is that blocks
deserve their own indented lines.

## 9. Read delimiter-related errors from the outside in

### Missing closing parenthesis

````python
total = (
    room_cost
    + materials_cost
````

Python may point at a later line or report that a parenthesis was never closed.
The cause is the unmatched `(`.

Repair:

````python
total = (
    room_cost
    + materials_cost
)
````

### Extra closing parenthesis

````python
total = room_cost + materials_cost)
````

There is no opening `(` for the final `)`. Remove it or add the intended opening
group.

### Missing comma between arguments

````python
print(
    "Course:"
    course_name,
)
````

The missing comma may be reported near `course_name`. Repair:

````python
print(
    "Course:",
    course_name,
)
````

### A string silently joins when a comma was intended

This code is valid:

````python
print(
    "Course:"
    "Python Foundations",
)
````

Adjacent string literals join, so the output is:

```text
Course:Python Foundations
```

If the intention was two arguments, add a comma:

````python
print(
    "Course:",
    "Python Foundations",
)
````

Valid syntax can still hide a punctuation mistake. Compare behavior with the
requirement, not only with the absence of an error.

### A closing delimiter is too early

````python
total = (
    room_cost
    + materials_cost
) + refreshment_cost
````

This is valid. The final addition happens outside the parenthesized group. If
the group is meant to show all event costs, move the closing delimiter:

````python
total = (
    room_cost
    + materials_cost
    + refreshment_cost
)
````

The numeric result is the same for addition, but the second layout communicates
the intended grouping.

## Key points

> **Key points**
- A logical statement can span several physical editor lines.
- Matching parentheses, brackets, or braces permit implicit continuation.
- Block indentation follows a colon-terminated header; continuation indentation
  arranges tokens inside open delimiters.
- Consistent vertical layout and trailing commas make structures easier to edit.
- Intermediate names improve meaning when line breaks alone are not enough.
- Prefer delimiter-based continuation over backslashes.
- Prefer one simple statement per line over semicolon-packed code.
- Pair delimiters before diagnosing the details inside a long statement.
