## 1. The values behind everyday Python programs

Unit 1 taught you to read Python's words and symbols, follow assignment, trace an
expression, recognize blocks, and repair syntax. You can now ask a deeper
question: **what kind of value should each name hold?**

Consider the beginning of a ticket for an imaginary observatory visit:

```python
visitor = "Mina"
group_size = 4
ticket_price = 7.50
has_evening_access = True
special_request = None
visit_date = "2026-08-14"
```

These values look simple, but each one makes a promise:

- `"Mina"` is text whose characters can be inspected and transformed.
- `4` is an exact count that can be divided into groups.
- `7.50` is a floating-point measurement-like number; exact money may need a
  different representation.
- `True` records a fact that can participate in a decision.
- `None` records that no special request was supplied.
- `"2026-08-14"` is still text. Python cannot safely do calendar arithmetic with
  it until the program parses it as a date.

This unit is about choosing those representations on purpose. A program becomes
more trustworthy when its values match the questions it must answer.

## 2. One spelling can describe different values

The text `"12"`, integer `12`, and floating-point number `12.0` may display
similarly, but Python gives them different operations and boundaries. The same is
true of an empty string, zero, `False`, and `None`: all can signal “nothing” in
casual conversation, yet they do not mean the same thing in a program.

You will learn to ask four questions whenever data enters a program:

1. **What does this value represent in the real task?**
2. **Which Python type supports the operations the task needs?**
3. **What information could conversion or rounding lose?**
4. **How should the program display the result without changing its meaning?**

Those questions connect every lesson in this unit.
