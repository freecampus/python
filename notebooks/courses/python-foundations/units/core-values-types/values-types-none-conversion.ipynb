## 1. The text `"12"` and the number `12` behave differently

A concert form sends a ticket quantity as text:

````python
quantity_text = "12"
print(quantity_text + quantity_text)
````

Before running it, decide whether the output will be `24`, `1212`, or an error.
Python prints `1212`. The `+` operator joins two strings because both values are
text. Quotation marks in source code are not decoration; they tell Python which
kind of value to create.

Compare the integer version:

````python
quantity = 12
print(quantity + quantity)
````

This time `+` performs arithmetic and prints `24`. The visible digits are the
same, but the **types** support different operations.

Run a wider comparison:

````python
whole = 12
measurement = 12.0
label = "12"
approved = True
missing = None

print(type(whole))
print(type(measurement))
print(type(label))
print(type(approved))
print(type(missing))
````

`type(value)` returns the value's type. Python displays `<class 'int'>`,
`<class 'float'>`, `<class 'str'>`, `<class 'bool'>`, and `<class 'NoneType'>`.
You do not need to memorize the angle-bracket display. The important names are
`int`, `float`, `str`, `bool`, and `NoneType`.

A name refers to a value. The value—not the spelling of the name—has a type.

### Ask for a compact type name

Types themselves have a `__name__` attribute:

````python
temperature = 21.5
print(type(temperature).__name__)
````

Python first evaluates `type(temperature)`, which produces the `float` type.
Then `.__name__` obtains the text name `"float"`. `print()` displays it without
the `<class ...>` wrapper.

### A value is not automatically converted to match an operation

Python refuses to guess how to combine unrelated types:

````python
ticket_count = 4
message = "Tickets: "
print(message + ticket_count)
````

The last line raises `TypeError`. The message explains that Python can concatenate
a string only with another string. Automatic guessing could conceal bad data, so
the program must state the intended conversion.

## 2. Inspect a category with `isinstance()`

`type()` answers “what is the direct type of this value?” Another built-in,
`isinstance()`, answers “does this value belong to this type or category?”

````python
distance = 7.5

print(type(distance).__name__)
print(isinstance(distance, float))
print(isinstance(distance, str))
````

The results are `float`, `True`, and `False`. The second argument to
`isinstance()` is a type, so it is written as `float` without quotes. The string
`"float"` would merely be text, not the type itself.

Use `type()` while investigating an unexpected value. Use `isinstance()` when a
program needs to ask whether a value belongs to an accepted type category.

> **A detail to recognize, not exploit**
`bool` is a specialized kind of integer in Python, so `isinstance(True, int)` is
`True` even though `type(True)` is `bool`. Write Boolean facts as `True` and
`False`; do not use that relationship as a shortcut for arithmetic. Lesson 7
examines Boolean boundaries in detail.

### Check the value you actually received

Predict all four results:

````python
participant_count = "8"

print(type(participant_count) is str)
print(isinstance(participant_count, str))
print(type(participant_count) is int)
print(isinstance(participant_count, int))
````

The first two are `True`; the last two are `False`. Both approaches agree for
this ordinary string. The code does **not** prove that the text contains a valid
participant count. Type inspection and content validation answer different
questions.

## 3. Conversion creates a result; it does not rewrite the source

Python's built-in type names can also act as conversion functions:

````python
quantity_text = "12"
quantity = int(quantity_text)

print(quantity_text, type(quantity_text).__name__)
print(quantity, type(quantity).__name__)
````

What Python does on each line:

1. `quantity_text = "12"` creates a string and binds the name.
2. `int(quantity_text)` reads the characters and creates the integer `12`.
3. Assignment binds the new integer to `quantity`.
4. Nothing reassigns `quantity_text`, so it still refers to the original string.

This distinction matters whenever the raw input is useful for error messages or
auditing. You can preserve the source and give the converted value a new name.

### Convert common numeric text

`int()` accepts optional surrounding whitespace and a sign:

````python
print(int("  42 "))
print(int("-7"))
print(int("+18"))
````

`float()` accepts decimal and scientific notation:

````python
print(float("3.75"))
print(float("-0.5"))
print(float("1.2e3"))
````

The final text means \(1.2 \times 10^3\), so Python prints `1200.0`.

### Convert a value for display

`str()` creates text:

````python
attempts = 3
attempts_text = str(attempts)

print("Attempts: " + attempts_text)
print(type(attempts).__name__)
print(type(attempts_text).__name__)
````

The integer remains an integer. The new string is useful for text concatenation,
although f-strings in Lesson 6 usually make mixed-value messages clearer.

### Parse a non-decimal integer when the contract says so

The optional second argument to `int()` states the numeral base:

````python
binary_flags = int("1011", 2)
hex_color_part = int("ff", 16)

print(binary_flags)
print(hex_color_part)
````

Python prints `11` and `255`. Do this only when the input contract actually says
binary or hexadecimal. `int("1011")` normally means the decimal number one
thousand eleven.

## 4. A successful conversion can still lose information

Conversion does not promise that every detail survives:

````python
measurement = 8.9
whole_part = int(measurement)

print(whole_part)
print(measurement)
````

`int(8.9)` produces `8`. It truncates toward zero; it does not round to the
nearest integer. The original `measurement` remains `8.9`, but `whole_part` no
longer records the fractional part.

Compare positive and negative values:

````python
print(int(3.9))
print(int(-3.9))
````

The results are `3` and `-3`. Both move toward zero. If a ticket rule says “always
use another seat for any partial group,” `int()` is the wrong operation. Lesson 2
will distinguish truncation, floor division, and rounding rules.

### Numeric conversions can change the available precision

````python
large_count = 10_000_000_000_000_001
approximate_count = float(large_count)
restored_count = int(approximate_count)

print(large_count)
print(approximate_count)
print(restored_count)
print(restored_count == large_count)
````

On standard Python, converting this large integer to a float cannot preserve every
digit. Converting it back does not recover the lost information. A round trip is
not evidence of safety unless the representation can express the original value.

### `bool()` answers truthiness, not human yes/no text

````python
print(bool("yes"))
print(bool("no"))
print(bool("False"))
print(bool(""))
````

The first three are `True` because every non-empty string is truthy. Only the empty
string is false. Do not use `bool(user_text)` to parse a human answer. Lesson 7
builds an explicit yes/no rule.

## 5. Read conversion failures before repairing them

Try this valid Python code in its own cell:

````python
participants = int("twelve")
````

Python raises `ValueError`. The call received the expected broad type—a string—but
the characters are not a valid base-ten integer literal.

Now try:

````python
participants = int(None)
````

This raises `TypeError`. `None` is not an accepted source for `int()`.

The distinction helps you describe the failure:

- **`ValueError`:** this kind of input is accepted, but this particular value
  cannot be interpreted as requested;
- **`TypeError`:** the operation does not accept this kind of value here.

Unit 8 will teach how to catch anticipated exceptions. For now, read the final
line, identify the conversion call, and fix either the source data or the program's
contract. Do not replace every failure with a made-up default.

### Decimal-looking text is not an integer literal

````python
whole = int("12.5")
````

This raises `ValueError`. If the contract permits decimal measurements, parse a
float first:

````python
measurement = float("12.5")
print(measurement)
````

Avoid writing `int(float("12.5"))` unless discarding the fractional part is an
explicit rule. Making an error disappear is not the same as preserving meaning.

### Test a conversion separately from a domain rule

````python
quantity_text = "-2"
quantity = int(quantity_text)

print(quantity)
print(quantity >= 0)
````

Conversion succeeds: `"-2"` is valid integer text. The comparison is `False`
because a negative ticket quantity violates this example's domain rule. Keep the
questions separate:

1. Can Python interpret the characters as an integer?
2. Does this program permit that integer?

## 6. `None` records deliberate absence

Suppose a registration form allows a nickname but does not require one:

````python
nickname = None
print(nickname is None)
````

`None` is a single special value used to represent “no value here” according to
the program's contract. It is not the text `"None"`, not zero, and not an empty
string.

Compare four possible states:

````python
unknown_score = None
zero_score = 0
blank_note = ""
literal_word = "None"

print(unknown_score is None)
print(zero_score is None)
print(blank_note is None)
print(literal_word is None)
````

Only the first result is `True`. Zero may be a real score. An empty string may mean
the person submitted a blank note. The word `"None"` is ordinary text.

### Check `None` with identity

Use:

````python
delivery_date = None
has_delivery_date = delivery_date is not None
print(has_delivery_date)
````

Python documentation and style guidance recommend `is None` and `is not None`.
Lesson 6 of the course later examines equality and identity in depth. For now,
treat this spelling as the standard `None` check.

### Do not convert absence into misleading text

````python
special_request = None
request_text = str(special_request)
print(request_text)
print(type(request_text).__name__)
````

The output looks like `None`, but `request_text` is now a four-character string.
It no longer represents absence. Convert to display text only at the display
boundary, and choose a human message deliberately:

````python
special_request = None
display_request = "No special request"
print(display_request)
````

Later decision lessons will select that display based on the value. The key idea
here is that representation carries meaning.

## Key points

> **Key points**
- A Python value has a type that determines which operations it supports.
- `type()` identifies a direct type; `isinstance()` asks whether a value belongs
  to a type category.
- Conversion creates a result and does not rewrite the source value.
- Successful conversion can discard information, so the domain must choose the
  rule.
- `ValueError` often signals unacceptable contents; `TypeError` often signals an
  unsupported kind of value.
- `None` represents deliberate absence and is normally checked with `is None`.
- Conversion and domain validation answer different questions.
