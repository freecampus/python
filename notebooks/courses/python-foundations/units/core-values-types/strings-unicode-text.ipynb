## 1. Quotation marks define a string in source code

Python strings hold text: names, labels, messages, identifiers, emoji, whitespace,
and line breaks. Create ordinary strings with single or double quotes:

````python
language = "Python"
greeting = 'Hello'

print(language)
print(greeting)
print(type(language).__name__)
````

The quote characters mark the beginning and end in source code. They are not part
of the displayed value.

Choose a quote style that avoids unnecessary escapes:

````python
message = "Mina's telescope is ready."
quotation = 'The display says "READY".'

print(message)
print(quotation)
````

Python treats single- and double-quoted strings the same after parsing them.
Readability determines the choice.

### Escape characters with a backslash

When the chosen quote appears inside the text, prefix it with `\`:

````python
message = 'Mina\'s telescope is ready.'
quotation = "The display says \"READY\"."

print(message)
print(quotation)
````

Common escape sequences include:

````python
two_lines = "north\nsouth"
tabbed = "name\tscore"
backslash = "C:\\observatory\\logs"

print(two_lines)
print(tabbed)
print(backslash)
````

`\n` is one newline character, `\t` is one tab character, and `\\` is one
literal backslash.

### Raw strings reduce backslash noise

A raw-string prefix makes most backslashes literal:

````python
pattern_text = r"\d{4}-\d{2}-\d{2}"
windows_folder = r"C:\observatory\logs"

print(pattern_text)
print(windows_folder)
````

Raw strings are useful for regular-expression patterns and Windows-style paths.
They are not a separate runtime type:

````python
print(type(r"raw text").__name__)
````

The answer is still `str`. A raw string also cannot end with one unmatched
backslash because that backslash would escape the closing quote in source code.

### Triple quotes hold multiline text

````python
instructions = """Observatory checklist:
1. Open the dome.
2. Check the weather.
3. Record the start time."""

print(instructions)
````

Line breaks inside the delimiters become part of the string. Triple-quoted strings
are also used as docstrings when they appear in a documented position. Unit 1
explained that placement, not the quote count alone, makes a docstring.

## 2. `len()` and indexes reveal a string's sequence

A string is an ordered sequence of Unicode code points. `len()` counts its current
sequence elements:

````python
word = "Python"

print(len(word))
print(word[0])
print(word[1])
print(word[-1])
````

The length is `6`. Index `0` selects `P`, index `1` selects `y`, and index `-1`
selects the final `n`.

Think of a nonnegative index as distance from the beginning:

```text
text:      P  y  t  h  o  n
index:     0  1  2  3  4  5
negative: -6 -5 -4 -3 -2 -1
```

### An index must exist

````python
word = "Python"
print(word[10])
````

Python raises `IndexError` because the only nonnegative indexes are `0` through
`5`. The error is evidence that the string is shorter than the program assumed.

An empty string has no valid character index:

````python
empty = ""
print(len(empty))
print(empty[0])
````

The first output is `0`; the second line raises `IndexError`.

### Slice a range of characters

A slice uses `start:stop`, including `start` and excluding `stop`:

````python
word = "Python"

print(word[0:2])
print(word[2:6])
print(word[:3])
print(word[3:])
````

The results are `Py`, `thon`, `Pyt`, and `hon`. Omitting a bound means “from the
beginning” or “through the end.”

Unlike a single index, a slice safely clips at the string boundary:

````python
word = "Python"
print(word[0:100])
````

The result is the whole word, not an error.

### Add a step to a slice

````python
alphabet = "abcdefghij"

print(alphabet[::2])
print(alphabet[1::2])
print(alphabet[::-1])
````

The third slice uses a step of `-1`, producing the characters in reverse order.
This is convenient for small text exercises; do not assume reversing code points
always preserves the way a human writing system should display.

## 3. Strings are immutable

Once created, a string's sequence of characters cannot be changed in place:

````python
word = "cat"
word[0] = "b"
````

Python raises `TypeError`: a string does not support item assignment. Build a new
string instead:

````python
word = "cat"
new_word = "b" + word[1:]

print(word)
print(new_word)
````

The original remains `"cat"`; the new value is `"bat"`.

String methods follow the same pattern:

````python
name = "  mina  "
cleaned_name = name.strip()

print(repr(name))
print(repr(cleaned_name))
````

`repr()` shows quotes and otherwise invisible surrounding spaces. `.strip()`
returns a new string; it does not update `name` automatically.

This common mistake discards the result:

````python
name = "  mina  "
name.strip()
print(repr(name))
````

The method ran, but no name kept its returned value. Reassign when that is the
intention:

````python
name = "  mina  "
name = name.strip()
print(repr(name))
````

````python
filename = "observations.csv"

print(".csv" in filename)
print(".json" not in filename)
````

Both results are Boolean values. Membership is case-sensitive:

````python
message = "Launch Ready"
print("ready" in message)
````

## 5. Clean boundaries with strip and prefix methods

External text often carries whitespace:

````python
raw_code = "\n  AX-42 \t"

print(repr(raw_code.strip()))
print(repr(raw_code.lstrip()))
print(repr(raw_code.rstrip()))
````

- `.strip()` removes whitespace from both ends;
- `.lstrip()` removes it from the left; and
- `.rstrip()` removes it from the right.

They do not remove whitespace from the middle:

````python
name = "Ada   Lovelace"
print(name.strip())
````

The internal spaces remain because they may be meaningful.

### Strip removes a set of edge characters, not a word

````python
label = "xy-report-yx"
print(label.strip("xy-"))
````

`.strip("xy-")` removes any of those individual characters repeatedly from both
edges. It does not look for the exact substring `"xy-"`.

Use explicit prefix/suffix methods when the contract names an exact marker:

````python
filename = "draft-observations.csv"

without_status = filename.removeprefix("draft-")
without_extension = without_status.removesuffix(".csv")

print(without_status)
print(without_extension)
````

If the exact prefix or suffix is absent, the method returns the original string.

### Empty and whitespace-only strings are different source values

````python
empty = ""
spaces = "   "

print(len(empty))
print(len(spaces))
print(spaces.strip() == "")
````

The raw lengths are `0` and `3`, but cleaning the spaces produces an empty string.
A form rule may treat both as blank while still preserving the raw source for
diagnosis.

## 6. Change case for display or comparison

Python provides several case transformations:

````python
label = "mOON BASE"

print(label.lower())
print(label.upper())
print(label.capitalize())
print(label.title())
````

These methods return new strings. `.title()` can be convenient for simple display
labels, but it is not a universal person-name formatter. Names and writing systems
have rules that a generic casing algorithm cannot infer.

For case-insensitive matching, `.casefold()` is more aggressive and
Unicode-aware than `.lower()`:

````python
first = "Straße"
second = "STRASSE"

print(first.lower() == second.lower())
print(first.casefold() == second.casefold())
````

The casefolded comparison is `True`. Keep the original strings for display; use a
normalized comparison form for matching.

## 7. Search, count, and replace deliberately

Use `in` for a yes/no membership question and `.find()` when the position matters:

````python
message = "weather station ready"

print("station" in message)
print(message.find("station"))
print(message.find("missing"))
````

`.find()` returns the starting index or `-1` when no match exists. Do not mistake
`-1` for a valid result position in later slicing.

If absence should be an error, `.index()` has a different contract:

````python
message = "weather station ready"
print(message.index("station"))
````

Calling `.index("missing")` would raise `ValueError`. Choose between a sentinel
result and an exception based on what absence means to the program.

### Count non-overlapping matches

````python
report = "rain rain sun rain"
print(report.count("rain"))
print(report.count("ai"))
````

The method counts non-overlapping occurrences. It does not understand words or
language boundaries; it searches exact character sequences.

### Test beginnings and endings

````python
filename = "observations.csv"

print(filename.startswith("observ"))
print(filename.endswith(".csv"))
````

These methods communicate intent more directly than manually slicing and
comparing a guessed number of characters.

### Replace exact character sequences

````python
status = "draft: telescope draft"
updated = status.replace("draft", "ready")

print(status)
print(updated)
````

Every non-overlapping match is replaced. Limit the number when the contract says
only the first occurrence:

````python
status = "draft: telescope draft"
print(status.replace("draft", "ready", 1))
````

Replacement is literal and case-sensitive; it is not a grammar-aware editing
tool.

## 8. Split text apart and join it again

`.split()` separates text and returns a list. Lists are the next unit, so treat the
container as supplied output here:

````python
raw_tags = "moon, stars, weather"
tags = raw_tags.split(",")

print(tags)
print(type(tags).__name__)
````

Spaces remain because the delimiter is only the comma. A later loop could clean
each item; for now, observe the boundary.

Splitting on no explicit separator treats runs of whitespace as one boundary:

````python
sentence = "  clear   dark sky  "
words = sentence.split()
print(words)
````

The result is `['clear', 'dark', 'sky']`.

`.join()` belongs to the delimiter string:

````python
words = ["clear", "dark", "sky"]
message = " | ".join(words)
print(message)
````

The output is `clear | dark | sky`. Every item must be a string:

````python
values = ["temperature", 21.5]
print(", ".join(values))
````

Python raises `TypeError` because joining does not guess how the float should be
formatted. Lesson 6 shows deliberate formatting.

## 9. Unicode text can have more than one code-point spelling

Python strings are Unicode text. `ord()` maps one character to its code point;
`chr()` maps a code point back:

````python
symbol = "A"
code_point = ord(symbol)

print(code_point)
print(chr(code_point))
print(ord("🌙"))
````

Code points are numeric identities in the Unicode standard. They are not the same
as encoded bytes; Unit 9 will teach UTF-8 and other encodings.

### Visually identical text can use different sequences

The character `é` can be one code point or an `e` followed by a combining accent:

````python
composed = "é"
decomposed = "e\u0301"

print(composed)
print(decomposed)
print(composed == decomposed)
print(len(composed))
print(len(decomposed))
````

The strings usually look alike, but equality is `False` and their lengths differ.
Normalize when the program's matching rule treats canonically equivalent Unicode
sequences as the same:

````python
import unicodedata

composed = "é"
decomposed = "e\u0301"

left = unicodedata.normalize("NFC", composed)
right = unicodedata.normalize("NFC", decomposed)

print(left == right)
````

NFC normalization converts both to a standard composed form where possible.

### Human-visible characters are not always one string element

Some emoji and writing-system clusters contain multiple code points joined for
display. `len()` counts Python string elements, not necessarily what a person
perceives as one symbol. Do not design cursor movement, text truncation, or user
name limits from `len()` alone without understanding the product's language and
display requirements.

### Normalize for matching, preserve originals for display

````python
import unicodedata

submitted = "  CAFÉ "
stored_label = "cafe\u0301"

submitted_key = unicodedata.normalize("NFC", submitted.strip()).casefold()
stored_key = unicodedata.normalize("NFC", stored_label).casefold()

print(submitted_key == stored_key)
print(submitted.strip())
````

The comparison succeeds while the original submitted capitalization remains
available. A comparison key is a tool, not a replacement for the person's chosen
display text.
