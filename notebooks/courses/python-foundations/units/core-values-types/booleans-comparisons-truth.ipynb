## 1. A comparison produces a value

A program often needs facts before it can make decisions. Comparisons produce the
two Boolean values `True` and `False`:

```python
temperature = 18

print(temperature >= 15)
print(temperature < 10)
print(type(temperature >= 15).__name__)
```

The outputs are `True`, `False`, and `bool`. A comparison is an expression, so its
result can be given a meaningful name:

```python
temperature = 18
minimum_temperature = 15

is_warm_enough = temperature >= minimum_temperature
print(is_warm_enough)
```

`is_warm_enough` exposes the question better than repeating a numeric expression
in every later output.

### Six ordinary comparison operators

| Operator | Meaning |
|---|---|
| `==` | equal values |
| `!=` | unequal values |
| `<` | less than |
| `<=` | less than or equal |
| `>` | greater than |
| `>=` | greater than or equal |

Do not confuse equality with assignment:

```python
score = 80
at_target = score == 80

print(score)
print(at_target)
```

The single `=` binds a name. The double `==` compares two values and produces a
Boolean result.

### Boundary symbols change the accepted set

```python
age = 18

print(age > 18)
print(age >= 18)
```

At exactly 18, the first comparison is `False` and the second is `True`. Translate
“older than” and “at least” carefully; one character changes the rule.

### Chain a range comparison

Python can express a closed range directly:

```python
score = 84
minimum = 60
maximum = 100

in_range = minimum <= score <= maximum
print(in_range)
```

This means `score` is at least `minimum` **and** at most `maximum`. It is usually
clearer than writing both comparisons separately.

Test the boundaries:

```python
print(60 <= 60 <= 100)
print(60 <= 100 <= 100)
print(60 <= 101 <= 100)
```

The inclusive endpoints are accepted; `101` is not.

## 2. Compare compatible meanings, not merely compatible syntax

Python can compare integers and floats numerically:

```python
print(4 == 4.0)
print(3 < 3.5)
```

Both are `True`. The types differ, but they participate in a common numeric
comparison.

Boolean values have a historical numeric relationship:

```python
print(True == 1)
print(False == 0)
print(True + True)
```

The results are `True`, `True`, and `2`. Recognize this behavior, but do not use
Boolean arithmetic when the program means a count. Give counts integer names and
facts Boolean names.

### String ordering is lexicographic by Unicode code points

```python
print("apple" < "banana")
print("Zoo" < "apple")
print("10" < "2")
```

Python compares the first differing character. Uppercase and lowercase code
points have their own order, and digit strings are compared as text. `"10" <
"2"` is `True` because the first character `"1"` comes before `"2"`.

Convert numeric text before a numeric comparison:

```python
left_text = "10"
right_text = "2"

print(int(left_text) < int(right_text))
```

The result is `False`, matching the numeric question.

### Unrelated ordering raises an error

```python
print("10" < 2)
```

Python raises `TypeError` rather than guessing whether to compare text or numbers.
Equality has a different contract and normally returns `False` for such unrelated
values:

```python
print("10" == 10)
```

That does not mean the values are “nearly equal.” It means they are not equal
under Python's value rules.

### Float comparisons need the precision contract

```python
import math

calculated = 0.1 + 0.2

print(calculated == 0.3)
print(math.isclose(calculated, 0.3))
```

Lesson 3 established the distinction: exact equality and closeness answer
different questions.

## 3. Truthiness asks whether a value counts as false in context

`bool(value)` reveals a value's truthiness:

```python
print(bool(0))
print(bool(7))
print(bool(-2))
print(bool(0.0))
```

Numeric zero is falsy. Other numbers are truthy, including negative numbers.

Strings follow emptiness:

```python
print(bool(""))
print(bool("hello"))
print(bool(" "))
print(bool("False"))
```

Only the empty string is falsy. A space is a character, and the word `"False"`
is non-empty text; both are truthy.

`None` is falsy:

```python
print(bool(None))
```

Collections use emptiness too. These are previews of Unit 3:

```python
print(bool([]))
print(bool([0]))
print(bool({}))
print(bool({"ready": False}))
```

An empty list or dictionary is falsy. A collection containing a falsy item is
still non-empty and therefore truthy.

### Truthiness can collapse different meanings

```python
unknown_score = None
zero_score = 0
blank_score = ""

print(bool(unknown_score))
print(bool(zero_score))
print(bool(blank_score))
```

All three are falsy, but they can mean “not measured,” “measured as zero,” and
“submitted as blank.” Use an explicit comparison when the distinction matters:

```python
score = 0

has_score = score is not None
score_is_positive = score > 0

print(has_score)
print(score_is_positive)
```

The program has a score, but it is not positive.

### Parse human Boolean text with an explicit rule

This is wrong for yes/no input:

```python
answer_text = "no"
answer = bool(answer_text)
print(answer)
```

It prints `True`. Instead, clean and compare the accepted spelling:

```python
answer_text = "  YES "
answer = answer_text.strip().casefold() == "yes"
print(answer)
```

This simple rule recognizes only `yes` as true; every other spelling produces
`False`. A production form may need to reject unknown answers rather than treating
them as no. Exception and validation units will design that boundary explicitly.

## 4. Combine facts with `not`, `and`, and `or`

`not` reverses truthiness:

```python
is_cloudy = False
is_clear = not is_cloudy

print(is_clear)
```

`and` requires both sides to be truthy:

```python
has_ticket = True
has_id = True

can_enter = has_ticket and has_id
print(can_enter)
```

`or` requires at least one side to be truthy:

```python
has_member_pass = False
has_day_ticket = True

has_admission = has_member_pass or has_day_ticket
print(has_admission)
```

### Name intermediate facts

```python
age = 17
has_permission = True
has_ticket = True

is_adult = age >= 18
age_requirement_met = is_adult or has_permission
can_enter = age_requirement_met and has_ticket

print(is_adult)
print(age_requirement_met)
print(can_enter)
```

The names show the rule in parts. A reader can inspect which fact changed instead
of decoding one long expression.

### Precedence is `not`, then `and`, then `or`

```python
result = True or False and False
print(result)
```

`and` runs first, so the expression is `True or False`, which is `True`.
Parentheses can state a different rule:

```python
result = (True or False) and False
print(result)
```

Now the result is `False`. Prefer named facts or parentheses whenever a reader
could reasonably group the rule another way.

### Short-circuiting can skip the right operand

For `and`, a falsy left side already determines the result:

```python
quantity = 0
safe_to_divide = quantity != 0 and 100 / quantity > 2

print(safe_to_divide)
```

Python does not evaluate `100 / quantity`, so no division-by-zero error occurs.

## 5. `and` and `or` return operands, not always Booleans

Try these expressions:

```python
print("Ada" and "ready")
print("" and "ready")
print("Ada" or "anonymous")
print("" or "anonymous")
```

The results are `"ready"`, `""`, `"Ada"`, and `"anonymous"`.

- `a and b` returns the first falsy operand, or the final operand if all are
  truthy.
- `a or b` returns the first truthy operand, or the final operand if all are
  falsy.

This explains the earlier optional-text pattern:

```python
coupon_text = "   "
coupon_code = coupon_text.strip() or None

print(coupon_code)
print(coupon_code is None)
```

The cleaned empty string is falsy, so `or` returns `None`.

Use this behavior for a deliberate fallback, not as a substitute for every
validation rule. A legitimate value such as zero would also trigger an `or`
fallback:

```python
recorded_score = 0
display_score = recorded_score or "not recorded"
print(display_score)
```

The output is misleading because zero is a real score. An explicit `is None`
decision is required; Unit 4 will implement it with a conditional.

## 6. Use `is` only for the identity question you mean

For missing values, use:

```python
review_date = None

print(review_date is None)
print(review_date is not None)
```

Do not use `is` as a substitute for equality with ordinary numbers or strings:

```python
left = "moon"
right = "".join(["m", "o", "o", "n"])

print(left == right)
print(left is right)
```

The values compare equal, but identity is an implementation-level object question
and may be false. Unit 6 will teach references and identity. Until then:

- use `==` and `!=` for ordinary value equality;
- use `is None` and `is not None` for the missing-value singleton.

## Key points

> **Key points**
- Comparisons produce Boolean values that can be named and inspected.
- Boundary operators such as `<` and `<=` express different accepted sets.
- String ordering compares characters, not numeric meaning.
- Truthiness collapses several empty or zero-like values; explicit comparisons
  preserve distinctions.
- `bool()` does not parse human yes/no text.
- `not`, `and`, and `or` combine facts and short-circuit evaluation.
- `and` and `or` return operands, so fallbacks must respect legitimate falsy
  values.
- Use `==` for ordinary value equality and `is None` for missing-value identity.
