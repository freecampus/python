## 9. Compare with a solution path

<details class="solution">
<summary>Reveal after all three hint levels have been used or your own checks pass</summary>

````python
import math
import unicodedata
from datetime import datetime, timedelta
from decimal import Decimal, ROUND_HALF_UP

traveler_text = "  aDA BYRON  "
item_text = "quantum compass"
quantity_text = "4"
unit_price_text = "3.25"
service_rate_text = "0.075"
mass_text = "0.30000000000000004"
priority_text = " YES "
coupon_text = "   "
purchased_at_text = "2026-08-03T14:30:00+00:00"
review_days_text = "30"

expected_mass = 0.3
mass_tolerance = 0.000001
cent = Decimal("0.01")

traveler = traveler_text.strip().title()
traveler_key = unicodedata.normalize("NFC", traveler_text.strip()).casefold()
item = item_text.strip().title()
quantity = int(quantity_text)
unit_price = Decimal(unit_price_text)
service_rate = Decimal(service_rate_text)
measured_mass = float(mass_text)
is_priority = priority_text.strip().casefold() == "yes"
coupon_code = coupon_text.strip() or None
purchased_at = datetime.fromisoformat(purchased_at_text)
review_days = int(review_days_text)

subtotal = unit_price * quantity
service_fee = (subtotal * service_rate).quantize(
    cent,
    rounding=ROUND_HALF_UP,
)
total = subtotal + service_fee
mass_matches = math.isclose(
    measured_mass,
    expected_mass,
    rel_tol=0.0,
    abs_tol=mass_tolerance,
)
mass_status = ("FAIL", "PASS")[mass_matches]
review_at = purchased_at + timedelta(days=review_days)
launch_code = (
    traveler[0].upper()
    + item[-1].upper()
    + purchased_at.strftime("%m%d")
)
record_ready = (
    quantity > 0
    and mass_matches
    and purchased_at.tzinfo is not None
    and coupon_code is None
)

summary = (
    "=== CHRONONAUT SUPPLY CLEARANCE ===\n"
    f"Traveler: {traveler} [{launch_code}]\n"
    f"Cargo: {quantity} × {item}\n"
    f"Credits: {subtotal:.2f} + {service_fee:.2f} = {total:.2f}\n"
    f"Mass check: {mass_status} | Priority: {is_priority}\n"
    f"Review: {review_at:%Y-%m-%d %H:%M UTC}"
)

print(summary)
````

This path keeps external text visible, performs calculations on typed values, and
creates display text only at the end. A different implementation is valid when it
preserves the names, constraints, and unchanged assertions.
