## 1. A console program is a small conversation

An interactive program follows a simple path:

1. display a useful prompt;
2. receive text from the person;
3. clean or convert that text;
4. perform the calculation; and
5. display a result with enough context to understand it.

`input()` always crosses into the program as text. Conversion is a separate,
visible step.

## 2. `print()` can display several values

`print()` accepts multiple arguments and inserts a space between them:

````python
name = "Mina"
score = 42

print("Observer", name, "scored", score)
````

Python converts each argument to display text for this output operation. The
integer `score` remains an integer afterward.

### Choose a separator with `sep`

````python
year = 2026
month = 8
day = 3

print(year, month, day, sep="-")
print("north", "east", "south", "west", sep=" | ")
````

`sep` is inserted **between** arguments, not after the final one.

### Choose what follows with `end`

`print()` normally ends with a newline. Override it deliberately:

````python
print("Loading", end="")
print(".", end="")
print(".", end="")
print(". done")
````

The output appears on one line. Use `end` sparingly; output that never moves to a
new line can be difficult to read in notebooks or terminals.

### Print an empty line

````python
print("Section A")
print()
print("Section B")
````

Calling `print()` with no arguments writes only its ending newline, creating one
blank line.

### Display is not storage

````python
quantity = 4
print("Quantity:", quantity)
print(type(quantity).__name__)
````

The first call displays a textual representation of `quantity`. It does not
replace the stored integer with text.

## 3. `input()` always returns a string

Run this cell and type `5`:

````python
number = input("Type a whole number: ")

print(number)
print(type(number).__name__)
print(number + number)
````

The type is `str`, and the final output is `55`. The keyboard sends characters;
Python does not infer a numeric type from their appearance.

Convert at a named boundary:

````python
number_text = input("Type a whole number: ")
number = int(number_text)
doubled = number * 2

print(f"Doubled: {doubled}")
````

The suffix `_text` makes the raw representation visible. `number` holds the
converted integer, and later arithmetic no longer needs to wonder whether the
value is text.

### Clean surrounding whitespace first

````python
city_text = input("City: ")
city = city_text.strip()

print(f"Raw: {city_text!r}")
print(f"Clean: {city!r}")
````

`!r` asks the f-string to use a diagnostic representation, so quotes and edge
spaces are visible. Keep both values while investigating input problems.

Numeric constructors already accept ordinary surrounding whitespace:

````python
quantity = int("  12  ")
measurement = float("  3.5  ")

print(quantity)
print(measurement)
````

Cleaning explicitly can still make the boundary and later empty checks easier to
explain.

## 4. F-strings place values inside readable messages

Prefix a string literal with `f` and place expressions inside braces:

````python
observer = "Mina"
objects_seen = 7

message = f"{observer} recorded {objects_seen} objects."
print(message)
````

Python evaluates each expression, converts its result for display, and constructs
one new string.

Forgetting the `f` leaves braces as ordinary characters:

````python
observer = "Mina"
print("Observer: {observer}")
````

The output is literally `Observer: {observer}`. Add the prefix:

````python
print(f"Observer: {observer}")
````

### Keep expressions understandable

Simple calculations can appear inside braces:

````python
completed = 8
remaining = 3

print(f"Total tasks: {completed + remaining}")
````

When an expression contains several operations or has meaning worth naming,
calculate it first:

````python
completed = 8
remaining = 3
total_tasks = completed + remaining

print(f"Total tasks: {total_tasks}")
````

The second version makes `total_tasks` available for checks and later output.

### Show the expression and value while debugging

The `=` debug form includes the expression text:

````python
temperature = 21.5
print(f"{temperature=}")
````

The output resembles `temperature=21.5`. This is useful while investigating, but
a user-facing report should usually have a human label.

### Use `!r` to reveal whitespace and escapes

````python
raw_name = "  Mina\n"

print(f"Human display: {raw_name}")
print(f"Diagnostic: {raw_name!r}")
````

The diagnostic representation shows quotes and `\n`. It is evidence for a
developer, not necessarily a friendly message for the person using the program.

## 5. Format specifications control appearance

Place a colon after the expression to add a format specification:

````python
distance = 12.34567
print(f"Distance: {distance:.2f} km")
````

`.2f` means fixed-point notation with two digits after the decimal point. The
f-string result is text; `distance` remains the original float.

### Add grouping separators

````python
visitors = 1250000
budget = 9876543.21

print(f"Visitors: {visitors:,}")
print(f"Budget: {budget:,.2f}")
````

The comma improves an English-style numeric display. Localization rules vary;
later applications may use a locale-aware presentation library.

### Display percentages

````python
completion = 0.875
print(f"Complete: {completion:.1%}")
````

Percentage formatting multiplies by 100 for display and adds `%`, producing
`87.5%`. The stored value remains `0.875`.

### Align fields into columns

````python
label = "Temperature"
value = 21.5

print(f"{label:<16} | {value:>8.2f}")
print(f"{'Humidity':<16} | {0.62:>8.1%}")
````

- `<16` left-aligns text in a field 16 characters wide;
- `>8.2f` right-aligns a two-decimal number in a field eight characters wide.

Fill and center a heading:

````python
print(f"{' NIGHT REPORT ':=^32}")
````

`=` is the fill character, `^` centers, and `32` is the field width.

### Display signs explicitly

````python
change = 4.2
loss = -1.7

print(f"Change: {change:+.1f}")
print(f"Change: {loss:+.1f}")
````

`+` in the specification displays a sign for both positive and negative values.

## 6. Human output and diagnostic output serve different readers

`str()` aims for a readable representation. `repr()` aims for an unambiguous or
developer-oriented representation when possible:

````python
message = "line one\nline two"

print(str(message))
print(repr(message))
````

The first output contains a real line break. The second includes quotes and the
visible escape `\n`.

F-strings use the same distinction:

````python
message = "line one\nline two"

print(f"User sees:\n{message}")
print(f"Developer inspects: {message!r}")
````

Do not show raw diagnostic representations as unexplained error messages to a
beginner user. Do use them while tracking invisible whitespace, newlines, and
unexpected types.

### Avoid formatting too early

````python
price = 3.25
price_text = f"{price:.2f}"

print(price_text)
print(type(price_text).__name__)
````

`price_text` is suitable for display but not for later numeric arithmetic. Keep
`price` for calculation and create display text at the output boundary.

## Key points

> **Key points**
- `input()` returns text, even when the characters look numeric.
- Clean and convert input at an explicit boundary before calculation.
- `print()` accepts multiple values and configurable `sep` and `end` text.
- F-strings produce strings and can include readable value expressions.
- Format specifications control precision, percentage, grouping, width, alignment,
  fill, and sign in the display.
- Formatting does not change the stored numeric value.
- Human-facing `str()` output and diagnostic `repr()` output serve different
  readers.
- Keep a fixed-data version of an interactive example so checks are reproducible.
