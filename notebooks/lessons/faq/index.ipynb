
### [Unexpected keyword arguments](unexpected-keyword-arguments-in-classes.qmd)

Understand why a class rejected a keyword, how calls reach `__init__`, and why
instance methods begin with `self`.

[Read the article](unexpected-keyword-arguments-in-classes.qmd){.card-link}

### [Running code and output](running-code-and-output.qmd)

Find out why nothing printed, why a notebook can show an old result, and when
to restart the runtime.

[Read the article](running-code-and-output.qmd){.card-link}

### [Names, values, and types](names-values-and-types.qmd)

Debug undefined variables, surprising assignments, text from `input()`, and
operations that combine incompatible types.

[Read the article](names-values-and-types.qmd){.card-link}

### No answer matches those words yet

Try a shorter symptom, or use the site search for a lesson covering the topic.

> **Search with the symptom**
Search for the words Python showed you and describe what happened, not only
what you hoped would happen. For example, search for `unexpected keyword
argument level`, `NameError undefined name`, or `input returns text` rather
than `my code is broken`.

## How to use this FAQ

Use this chapter when Python does something surprising and you are not sure
which article can help. Begin with the words that describe what you see, such as
"unexpected keyword argument," "nothing printed," "this name is undefined,"
or "Python will not add these values."

Each FAQ page is a standalone article: start with the short answer, then read as
much of the explanation as you need. If you want to experiment, use its **Open
in Colab** link and follow this optional routine:

1. Find the symptom that resembles the problem.
2. Predict one likely cause before applying a fix.
3. Run the smallest example and inspect the evidence.
4. Change one thing, then run the example again.
5. Explain what the result showed in one sentence.

While experimenting, keep one small troubleshooting note with four parts:
**symptom**, **clue**, **change**, and **result**. You can stop when the article
answers your question or continue into the linked course lessons for structured
practice.

## Questions answered elsewhere

This FAQ provides short diagnostic paths. Use the main lessons when you need a
complete explanation and more practice.

| If you need help with... | Continue with... |
| --- | --- |
| A traceback or named exception | [Syntax Errors, Runtime Exceptions, and Logic Bugs](../../courses/python-foundations/units/errors-exceptions-debugging/failure-categories-tracebacks.qmd) |
| Inspecting a program while it runs | [A Repeatable Debugging Method](../../courses/python-foundations/units/errors-exceptions-debugging/repeatable-debugging-method.qmd) |
| Making a small example to share | [Debugger Tools and Minimal Reproducible Examples](../../courses/python-foundations/units/errors-exceptions-debugging/debugger-minimal-reproductions.qmd) |
| Classes, instances, methods, and `self` | [Build Your First Useful Class](../../courses/python-foundations/units/object-oriented-python/objects-classes-instances-self.qmd) |
| Variables and Python types | [Values, Names, and Assignment](../../courses/python-foundations/units/python-syntax/values-names-assignment.qmd) |
| Text input and printed output | [Input, Output, and Clear Formatting](../../courses/python-foundations/units/core-values-types/input-output-formatting.qmd) |
| Colab cells and runtimes | [Work Reliably in Colab and Jupyter](../../courses/python-foundations/units/get-started/notebook-workflows.qmd) |

[Return to the full lesson path](../../courses/index.qmd) when the immediate problem is
solved.

## Choose the next useful step

Use the evidence in each situation to choose one small diagnostic step.

## Keep the FAQ useful

When you encounter a repeated question that is not covered here, record:

- the exact symptom or error message;
- the smallest code that shows it;
- the clue that identified the cause;
- the smallest change that fixed it; and
- the lesson that explains the underlying concept.

Questions with the same theme can become a cohesive FAQ article. A single
question that already has a full course lesson should become a short link rather
than duplicated material.
